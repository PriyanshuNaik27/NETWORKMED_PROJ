```python
drug_target_intertaction = pd.read_csv('../raw_data/drug.target.interaction.tsv.gz',sep="\t", compression="gzip")
```

```python
drug_target_intertaction.sample(5)
```

we only need homo sapiens 


```python
drug_target_human = drug_target_intertaction[drug_target_intertaction['ORGANISM']=='Homo sapiens']
drug_target_human.head()
```

```python
promiscuity_df =    drug_target_human.groupby('DRUG_NAME')['GENE'].nunique().reset_index()
promiscuity_df.head()
```

```python
promiscuity_df.columns = ['Drug_name','Promiscuity_index']
promiscuity_df.head()
```

```python
promiscuity_df['Drug_name'] = promiscuity_df['Drug_name'].str.lower()
```

```python

promiscuity_df.sort_values(by='Promiscuity_index', ascending=False).to_csv('../processed_data/promiscuity_index.csv', index=False)
```
